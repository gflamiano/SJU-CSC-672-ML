## HW 2 Analysis

**1. What challenges did you encounter when loading and inspecting a raw external dataset from Kaggle compared to pre-packaged datasets?**  
One challenge was that the dataset was a raw CSV file, so I had to check the data myself to understand the columns, data types, and overall structure. With pre-packaged datasets, this information is usually easier to access. One must use `head()`, `info()`, and `describe()` to first inspect the dataset and get a better understanding of the numerical and categorical features.

**2. How did you determine which columns required missing value imputation versus other data cleaning methods, and what strategy did you choose?**  
I first checked the columns using `isna().sum()` and calculated the percentage of missing values. The results showed that there were no actual missing values in the dataset. I also noticed that `"None"` in the `encryption_used` column was a valid category, so I made sure it was not incorrectly treated as a missing value. I still included median imputation for numerical columns and most frequent imputation for categorical columns in the preprocessing pipeline in case missing values are present in future data.

**3. Explain how the Scikit-Learn `ColumnTransformer` and `Pipeline` streamline the data preparation process and prevent data leakage across preprocessing steps.**  
The `ColumnTransformer` makes it easier to use different preprocessing methods for different types of columns. For example, numerical features can be scaled with `StandardScaler`, while categorical features can be converted using `OneHotEncoder`. The `Pipeline` combines these steps so the preprocessing can be applied in an organized and consistent way. It also helps prevent data leakage by making sure the preprocessing is learned from the training data before being applied to the test data.

**4. Based on your correlation matrix and heatmap, which features show the strongest relationships with the target variable (`attack_detected`), and what operational insights do these relationships reveal?**  
Looking at the correlation matrix and heatmap, `login_attempts` and `failed_logins` had some of the strongest relationships with `attack_detected` among the numerical features. This makes sense because a large number of login attempts or failed logins could indicate suspicious behavior, such as repeated attempts to gain access. These features could be useful when trying to identify potential attacks.
